## 作業說明

本作業使用 Python 整理「桌遊體驗」與「電影欣賞」兩場活動的報名資料。

本作業會使用字串方法、list、tuple、dict、set，
完成報名資料整理、去除重複、集合比較、人數統計與費用計算。

```python
# HW09 活動報名整理器
# 建立原始資料

# 固定的兩個活動名稱，用 tuple 保存
activity_names = ("桌遊體驗", "電影欣賞")

# 原始報名文字，保留原本內容
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 每個活動的每人費用，用 dict 保存
unit_fees = {
    "桌遊體驗": 120,
    "電影欣賞": 80
}

print("姓名：張庭瑄")
print("學號：D1370904")
print("活動名稱：", activity_names)
print("桌遊原始資料：", board_raw)
print("電影原始資料：", movie_raw)
print("活動費用：", unit_fees)
```

```python
# 從 tuple 中取得兩個活動名稱
board_activity, movie_activity = activity_names

print("桌遊活動名稱：", board_activity)
print("電影活動名稱：", movie_activity)
```

```python
# 整理桌遊原始文字
board_list = board_raw.replace(" ", "").lower().split(",")

# 整理電影原始文字
movie_list = movie_raw.replace(" ", "").lower().split(",")

print("桌遊整理串列：", board_list)
print("電影整理串列：", movie_list)
```

```python
# 將兩場活動的 list 轉成 set
# set 會自動去除重複的人

board_set = set(board_list)
movie_set = set(movie_list)

print("桌遊去重集合：", board_set)
print("電影去重集合：", movie_set)
```

```python
# 兩場都有參加的人
common_set = board_set & movie_set

# 至少參加一場的人
all_set = board_set | movie_set

# 只參加桌遊的人
board_only_set = board_set - movie_set

# 只參加電影的人
movie_only_set = movie_set - board_set
```

```python
# 將集合排序成 list
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
common_sorted = sorted(common_set)
board_only_sorted = sorted(board_only_set)
movie_only_sorted = sorted(movie_only_set)
all_sorted = sorted(all_set)

print("桌遊去重名單：", board_sorted)
print("電影去重名單：", movie_sorted)
print("共同參加者：", common_sorted)
print("只參加桌遊者：", board_only_sorted)
print("只參加電影者：", movie_only_sorted)
print("全部參加者：", all_sorted)
```

```python
# 從排序後的全部參加者中取得前兩位
preview = all_sorted[:2]

print("通知名單前兩位：", preview)
```

```python
# 計算各種人數
board_count = len(board_set)
movie_count = len(movie_set)
common_count = len(common_set)
total_unique_count = len(all_set)

# 報名人次 = 兩場各自去重後的人數相加
signup_count = board_count + movie_count

print("桌遊人數：", board_count)
print("電影人數：", movie_count)
print("共同人數：", common_count)
print("總不重複人數：", total_unique_count)
print("報名人次：", signup_count)
```

```python
# 從 unit_fees 查詢兩場活動的每人費用
board_fee = unit_fees[board_activity]
movie_fee = unit_fees[movie_activity]

# 使用「去重後的人數」計算費用
board_total = board_count * board_fee
movie_total = movie_count * movie_fee

# 兩場加總
total_fee = board_total + movie_total

print("桌遊費用：", board_total)
print("電影費用：", movie_total)
print("總費用：", total_fee)
```

```python
# 建立摘要字典
summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": common_count,
    "總不重複人數": total_unique_count,
    "報名人次": signup_count,
    "桌遊應收": board_total,
    "電影應收": movie_total,
    "總應收": total_fee
}

print("摘要：", summary)
```

```python
print("HW09 活動報名摘要")

print(f"活動：{board_activity}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_activity}，每人 {movie_fee:.2f} 元")

print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")

print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")

print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")

print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['共同人數']} 人")

print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")

print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")

print(f"通知名單前兩位：{preview}")

print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
